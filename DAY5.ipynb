```python
salary = np.array([
    34000,78000,53400,87730,738380,733830,452446,49884,93987,93794,
    98480,89767,76789,898868,87878,897979,99978,78303,983033,983093,
    380383,99484,9099393,7887983,3254346,5587686,86879,68698,78889,
    98033,9397393,892727,83983,398739,83733,90933
])
print("salaries :" , salary)
print("highest:" , max(salary))
print("lowest:" , min(salary))
print("average:" , sum(salary)/len(salary))

salary.sort()
print("sorted salary:" , salary)

increment =  0.1* salary
newsalary=  increment+ salary
print("after increment:" , newsalary)

print("increase:" , newsalary-salary)











```
