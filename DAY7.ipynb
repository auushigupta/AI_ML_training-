```python
#linear regression

from sklearn.linear_model import LinearRegression
 #house size
X = [[1000],[1200],[1500],[1800]]
#HOUSE PRIZE in lakhs
y = [40,50,70,100]

model = LinearRegression()
model.fit(X,y)

prediction = model.predict([[1600]])

print("predicted prize:" ,prediction[0], "lakh")
```

```python
#clustering k means
from sklearn.cluster import KMeans
import numpy as np

data = np.array([
    [2,3],
    [3,4],
    [10,12],
    [11,13]
])
model = KMeans(n_clusters = 2,random_state = 42)
model.fit(data)

print("Cluster Labels:")
print(model.labels_)
```
