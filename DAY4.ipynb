```python
class fruits:
    def seed(self):
        print("has seeds")

class citrusfruits(fruits) :
     def sour(self) :
         print("is sour")

     def peel(self):
          print("has peel")

f = citrusfruits()

f.sour()
f.seed()
f.peel()
```
